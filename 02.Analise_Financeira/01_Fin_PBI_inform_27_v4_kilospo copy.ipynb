```python
# ==========================
# 46. Preparar data
# ==========================
df_danone['data_entrega'] = pd.to_datetime(
    df_danone['PFEENT']
    .astype('string')
    .str.strip(),
    format='%Y%m%d',
    errors='coerce'
)

# ==========================
# 47. Calcular total mensal
# ==========================
ingresso_por_mes = (
    df_danone
    .assign(
        mes=df_danone['data_entrega']
        .dt.to_period('M')
        .astype('string')
    )
    .groupby(
        'mes',
        dropna=False
    )
    .agg(
        registos=('PREFPE', 'size'),
        valores_calculados=(
            'ingresso_calculado',
            lambda x: x.notna().sum()
        ),
        valores_nulos=(
            'ingresso_calculado',
            lambda x: x.isna().sum()
        ),
        ingresso_total=(
            'ingresso_calculado',
            lambda x: x.sum(min_count=1)
        )
    )
    .reset_index()
    .sort_values('mes')
)

# ==========================
# 48. Adicionar total
# ==========================
linha_total = pd.DataFrame({
    'mes': ['TOTAL'],
    'registos': [
        ingresso_por_mes['registos'].sum()
    ],
    'valores_calculados': [
        ingresso_por_mes['valores_calculados'].sum()
    ],
    'valores_nulos': [
        ingresso_por_mes['valores_nulos'].sum()
    ],
    'ingresso_total': [
        ingresso_por_mes['ingresso_total'].sum(
            min_count=1
        )
    ]
})

ingresso_por_mes = pd.concat(
    [
        ingresso_por_mes,
        linha_total
    ],
    ignore_index=True
)

# ==========================
# 49. Formatar números
# ==========================
def formatar_numero(valor):
    if pd.isna(valor):
        return ''

    return (
        f'{valor:,.1f}'
        .replace(',', 'X')
        .replace('.', ',')
        .replace('X', '.')
    )

# ==========================
# 50. Mostrar resumo
# ==========================
ingresso_por_mes.style.format({
    'registos': formatar_numero,
    'valores_calculados': formatar_numero,
    'valores_nulos': formatar_numero,
    'ingresso_total': formatar_numero
})
```

```python
# ==========================
# 46. Preparar Danone
# ==========================
df_ingresso_danone = df_danone[
    [
        'PREFPE',
        'PFEENT',
        'ingresso_danone'
    ]
].copy()

df_ingresso_danone['PREFPE'] = (
    df_ingresso_danone['PREFPE']
    .astype('string')
    .str.strip()
)

df_ingresso_danone['PFEENT'] = pd.to_datetime(
    df_ingresso_danone['PFEENT']
    .astype('string')
    .str.strip(),
    format='%Y%m%d',
    errors='coerce'
)

# ==========================
# 47. Somar por entrega
# ==========================
df_ingresso_danone = (
    df_ingresso_danone
    .groupby(
        [
            'PREFPE',
            'PFEENT'
        ],
        dropna=False
    )
    .agg(
        ingresso_danone_total=(
            'ingresso_danone',
            lambda x: x.sum(min_count=1)
        )
    )
    .reset_index()
)

# ==========================
# 48. Preparar inform
# ==========================
df['REFERENCIA'] = (
    df['REFERENCIA']
    .astype('string')
    .str.strip()
)

df['FENTREGA'] = pd.to_datetime(
    df['FENTREGA'],
    errors='coerce'
)

df['_ordem_original'] = range(len(df))

# ==========================
# 49. Remover resultado anterior
# ==========================
colunas_anteriores = [
    coluna
    for coluna in [
        'ingresso_danone',
        'ingresso_danone_total'
    ]
    if coluna in df.columns
]

df = df.drop(
    columns=colunas_anteriores
)

# ==========================
# 50. Associar ingresso
# ==========================
df = df.merge(
    df_ingresso_danone,
    left_on=[
        'REFERENCIA',
        'FENTREGA'
    ],
    right_on=[
        'PREFPE',
        'PFEENT'
    ],
    how='left',
    validate='many_to_one'
)

df = df.sort_values(
    '_ordem_original'
).reset_index(drop=True)

# ==========================
# 51. Identificar última linha
# ==========================
ultima_linha = ~df.duplicated(
    subset=[
        'REFERENCIA',
        'FENTREGA'
    ],
    keep='last'
)

# ==========================
# 52. Atribuir ingresso
# ==========================
df['ingresso_danone'] = pd.Series(
    pd.NA,
    index=df.index,
    dtype='Float64'
)

df.loc[
    ultima_linha,
    'ingresso_danone'
] = df.loc[
    ultima_linha,
    'ingresso_danone_total'
]

# ==========================
# 53. Limpar colunas auxiliares
# ==========================
df = df.drop(
    columns=[
        'PREFPE',
        'PFEENT',
        'ingresso_danone_total',
        '_ordem_original'
    ]
)

# ==========================
# 54. Calcular total
# ==========================
df['INGRESODT'] = pd.to_numeric(
    df['INGRESODT'],
    errors='coerce'
)

df['total_ingresso'] = (
    df['INGRESODT'].fillna(0)
    + df['ingresso_danone'].fillna(0)
)

# ==========================
# 55. Validar total
# ==========================
validacao_ingresso = pd.DataFrame({
    'indicador': [
        'Ingresso no df_danone',
        'Ingresso associado ao df',
        'Diferença'
    ],
    'valor': [
        df_danone['ingresso_danone'].sum(min_count=1),
        df['ingresso_danone'].sum(min_count=1),
        (
            df_danone['ingresso_danone'].sum(min_count=1)
            - df['ingresso_danone'].sum(min_count=1)
        )
    ]
})

validacao_ingresso
```

```python
# ==========================
# 56. Preparar chaves do df
# ==========================
chaves_df = (
    df[
        [
            'REFERENCIA',
            'FENTREGA'
        ]
    ]
    .drop_duplicates()
    .copy()
)

# ==========================
# 57. Identificar não associados
# ==========================
ingressos_sem_correspondencia = (
    df_ingresso_danone
    .merge(
        chaves_df,
        left_on=[
            'PREFPE',
            'PFEENT'
        ],
        right_on=[
            'REFERENCIA',
            'FENTREGA'
        ],
        how='left',
        indicator=True,
        validate='one_to_one'
    )
    .query('_merge == "left_only"')
    .drop(
        columns=[
            'REFERENCIA',
            'FENTREGA',
            '_merge'
        ]
    )
    .sort_values(
        'ingresso_danone_total',
        ascending=False
    )
)

# ==========================
# 58. Resumir diferenças
# ==========================
resumo_sem_correspondencia = pd.DataFrame({
    'indicador': [
        'Entregas sem correspondência',
        'Entregas com ingresso',
        'Entregas com ingresso nulo',
        'Ingresso sem correspondência'
    ],
    'valor': [
        len(ingressos_sem_correspondencia),
        ingressos_sem_correspondencia[
            'ingresso_danone_total'
        ].notna().sum(),
        ingressos_sem_correspondencia[
            'ingresso_danone_total'
        ].isna().sum(),
        ingressos_sem_correspondencia[
            'ingresso_danone_total'
        ].sum(min_count=1)
    ]
})

# ==========================
# 59. Formatar resumo
# ==========================
resumo_sem_correspondencia.style.format({
    'valor': formatar_numero
})
```

```python
# ==========================
# 61. Preparar referências do df
# ==========================
referencias_df = (
    df
    .groupby(
        'REFERENCIA',
        dropna=False
    )
    .agg(
        primeira_data_df=('FENTREGA', 'min'),
        ultima_data_df=('FENTREGA', 'max'),
        datas_df=('FENTREGA', 'nunique')
    )
    .reset_index()
)

# ==========================
# 62. Verificar referências
# ==========================
analise_sem_correspondencia = (
    ingressos_sem_correspondencia
    .merge(
        referencias_df,
        left_on='PREFPE',
        right_on='REFERENCIA',
        how='left',
        validate='many_to_one'
    )
)

# ==========================
# 63. Classificar motivo
# ==========================
analise_sem_correspondencia['motivo'] = (
    'REFERENCIA_EXISTE_DATA_DIFERENTE'
)

analise_sem_correspondencia.loc[
    analise_sem_correspondencia[
        'REFERENCIA'
    ].isna(),
    'motivo'
] = 'REFERENCIA_INEXISTENTE'

analise_sem_correspondencia['mes'] = (
    analise_sem_correspondencia['PFEENT']
    .dt.to_period('M')
    .astype('string')
)

# ==========================
# 64. Resumir motivos
# ==========================
resumo_motivos = (
    analise_sem_correspondencia
    .groupby(
        [
            'mes',
            'motivo'
        ],
        dropna=False
    )
    .agg(
        entregas=('PREFPE', 'size'),
        entregas_com_ingresso=(
            'ingresso_danone_total',
            lambda x: x.notna().sum()
        ),
        ingresso_total=(
            'ingresso_danone_total',
            lambda x: x.sum(min_count=1)
        )
    )
    .reset_index()
    .sort_values(
        [
            'mes',
            'motivo'
        ]
    )
)

# ==========================
# 65. Mostrar resumo
# ==========================
resumo_motivos.style.format({
    'entregas': formatar_numero,
    'entregas_com_ingresso': formatar_numero,
    'ingresso_total': formatar_numero
})
```

```python
# Garantir que capacidade_norm existe
if 'capacidade_norm' not in df.columns:
    print("Criando capacidade_norm...")
    
    # Converter para número
    df['CAMION_CAPACIDAD_NUM'] = pd.to_numeric(df['CAMION_CAPACIDAD'], errors='coerce')
    
    # Mapeamento de normalização
    mapeamento = {
        4: 6, 5: 6, 6: 6,
        8: 12, 12: 12,
        14: 20, 15: 20, 16: 20, 18: 20, 20: 20,
        22: 24, 24: 24,
        33: 33, 66: 66
    }
    
    # Aplicar mapeamento
    df['capacidade_norm'] = df['CAMION_CAPACIDAD_NUM'].map(mapeamento)
    
    # Tratar zeros
    linhas_zero = df['CAMION_CAPACIDAD_NUM'] == 0
    if linhas_zero.sum() > 0:
        capacidade_por_codeut = df[df['CAMION_CAPACIDAD_NUM'] > 0].groupby('CODEUT')['capacidade_norm'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else x.max())
        for cu in df[linhas_zero]['CODEUT'].unique():
            if cu in capacidade_por_codeut.index:
                df.loc[(df['CODEUT'] == cu) & linhas_zero, 'capacidade_norm'] = capacidade_por_codeut[cu]

# Criar coluna indicadora
df['dados_validos'] = df['capacidade_norm'].notna()

print("="*80)
print("MARCAÇÃO DE DADOS")
print("="*80)

print(f"Total linhas: {len(df):,}")
print(f"Dados válidos: {df['dados_validos'].sum():,}")
print(f"Dados incompletos (ignorar): {(~df['dados_validos']).sum():,}")

print(f"\n--- Cálculos usarão apenas linhas com dados_validos = True ---")
```
