```python
import pandas as pd

# Célula 1: Ler o ficheiro
df = pd.read_csv(r'C:\Users\LISARR\Documents\python\00.Metadados\CP7COORDS.CSV', sep=';')
df.head()
```

```python
# Célula 2: Criar a coluna CP com formatting condicional (SEM "CP ")
df['CP'] = df['CP4'].astype(str) + '-' + df['CP3'].astype(str).str.zfill(3)
df[['CP', 'POINT_X', 'POINT_Y']].head()
```

```python
# Célula 3: Ver resultado final
df[['CP', 'POINT_X', 'POINT_Y']].head()
```

```python
# Célula 1: Conectar à base de dados
import sqlite3
DB_PATH = r"C:\Users\LISARR\Documents\python\00.DB\2026.db"
con = sqlite3.connect(DB_PATH)
print("✓ Conectado à base de dados")
```

```python
# Célula 2: Criar tabela
cursor = con.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS coordenadas (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        cp TEXT NOT NULL,
        point_x REAL,
        point_y REAL
    )
''')

con.commit()
print("✓ Tabela 'coordenadas' criada")
```

```python
# Célula 3c: Selecionar e renomear colunas
df_inserir = df[['CP', 'POINT_X', 'POINT_Y']].copy()
df_inserir.columns = ['cp', 'point_x', 'point_y']
df_inserir.head()
```

```python
# Célula 3d: Inserir dados na BD
df_inserir.to_sql('coordenadas', con, if_exists='append', index=False)
con.commit()

print(f"✓ {len(df_inserir)} registos inseridos")
```
