```python

# ==========================
# 1. Bibliotecas
# ==========================
from pathlib import Path
import pandas as pd

# ==========================
# 2. Configuração
# ==========================
PASTA = Path.home() / "Downloads" / "OneDrive_2_10-11-2026"

# ==========================
# 3. Ler e juntar ficheiros
# ==========================
ficheiros = sorted(PASTA.glob("*.xlsx"))

df = pd.concat(
    [pd.read_excel(ficheiro) for ficheiro in ficheiros],
    ignore_index=True
)

# ==========================
# 4. Resultado
# ==========================
display(df)

print(f"Ficheiros lidos: {len(ficheiros)}")
print(f"Total de linhas: {len(df):,}")
print(f"Total de colunas: {len(df.columns)}")
```

```python
df.to_excel(Path.home() / "Documents" / "dados_compilados.xlsx", index=False)
```
