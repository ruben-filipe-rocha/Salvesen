```python
# Agrupar por TRANSPORTISTA (soma todas as linhas de entregas)
df_transp = df.groupby('TRANSPORTISTA').agg({
    'COSTEUT': 'sum',
    'PALESREAL': 'sum'
})

# Converter PALESREAL para numérico
df_transp['PALESREAL'] = pd.to_numeric(df_transp['PALESREAL'], errors='coerce')

# Calcular custo por palete
df_transp['custo_palete'] = (df_transp['COSTEUT'] / df_transp['PALESREAL']).round(2)

# Ordenar por custo_palete
df_transp = df_transp.sort_values('custo_palete', ascending=False)

# Formatar para apresentação
df_transp_display = df_transp.copy()
df_transp_display['COSTEUT'] = df_transp_display['COSTEUT'].apply(lambda x: f"{x:,.2f}")
df_transp_display['PALESREAL'] = df_transp_display['PALESREAL'].apply(lambda x: f"{x:,.2f}")
df_transp_display['custo_palete'] = df_transp_display['custo_palete'].apply(lambda x: f"{x:,.2f}")

print(f"Transportadores: {len(df_transp)}")
df_transp_display
```

```python
# Caminho do ficheiro
caminho = r"C:\Users\LISARR\Documents\python\01.Financeiro\Danone_Custos_V2.xlsx"

# Ler sheet "Ingreso y coste" colunas 0-4 (A-E), ignorando primeira linha
df_danone = pd.read_excel(
    caminho,
    sheet_name="Ingreso y coste",
    usecols=range(5),
    skiprows=1
)

print(f"Shape: {df_danone.shape}")
df_danone.head()
```

```python
df_danone.columns.tolist()
```

```python
# Limpar espaços/caracteres não numéricos

# Remover linha de cabeçalho "Etiquetas de fila" se existir
df_danone_clean = df_danone[df_danone['CODEDT'] != 'Etiquetas de fila'].copy()

print(f"Registos df_danone_clean: {len(df_danone_clean)}")
df_danone_clean.head()
df_danone_clean['CODEDT_numeros_limpo'] = df_danone_clean['CODEDT'].astype(str).str.replace(r'\D', '', regex=True)
df['REFERENCIA_limpo'] = df['REFERENCIA'].astype(str).str.replace(r'\D', '', regex=True)

# Primeira sequência (antes do espaço original no Excel)
df_danone_clean['primeira_seq_numeros'] = df_danone_clean['CODEDT'].astype(str).str.split().str[0]
df_danone_clean['primeira_seq_numeros'] = df_danone_clean['primeira_seq_numeros'].str.replace(r'\D', '', regex=True)

# Set para procura rápida
referencias_set = set(df['REFERENCIA_limpo'].dropna())

# Estratégia 1: sequência completa
df_danone_clean['match_tipo'] = 'sem_match'
match_completo = df_danone_clean['CODEDT_numeros_limpo'].isin(referencias_set)
df_danone_clean.loc[match_completo, 'match_tipo'] = 'completo'

# Estratégia 2: primeira sequência (só para os que falharam na 1)
sem_match_mask = df_danone_clean['match_tipo'] == 'sem_match'
match_primeira = df_danone_clean.loc[sem_match_mask, 'primeira_seq_numeros'].isin(referencias_set)
df_danone_clean.loc[sem_match_mask & match_primeira.reindex(df_danone_clean.index, fill_value=False), 'match_tipo'] = 'primeira_seq'

print(f"Matches com sequência completa: {(df_danone_clean['match_tipo'] == 'completo').sum()}")
print(f"Matches com primeira sequência: {(df_danone_clean['match_tipo'] == 'primeira_seq').sum()}")
print(f"Sem match: {(df_danone_clean['match_tipo'] == 'sem_match').sum()}")
print(f"Taxa total de match: {(df_danone_clean['match_tipo'] != 'sem_match').sum() / len(df_danone_clean) * 100:.2f}%")
```

```python
# Criar chave de match única (usa sequência completa, senão primeira sequência)
df_danone_clean['match_key'] = df_danone_clean.apply(
    lambda row: row['A_numeros_limpo'] if row['match_tipo'] == 'completo' 
    else (row['primeira_seq_numeros'] if row['match_tipo'] == 'primeira_seq' else None),
    axis=1
)

# Agrupar Danone por match_key (somar Ingreso = coluna D, caso haja duplicados)
danone_ingreso = df_danone_clean.dropna(subset=['match_key']).groupby('match_key')['D'].sum().reset_index()
danone_ingreso.columns = ['match_key', 'Ingreso']

# Merge no df
df = df.merge(
    danone_ingreso,
    left_on='REFERENCIA_limpo',
    right_on='match_key',
    how='left'
)
df = df.drop(columns=['match_key'])

# Preencher NaN com 0 (linhas sem correspondência Danone)
df['Ingreso'] = df['Ingreso'].fillna(0)

# Zerar Ingreso nas linhas duplicadas de REFERENCIA (cliente só paga 1x por REFERENCIA)
primeira_ocorrencia = ~df['REFERENCIA_limpo'].duplicated()
df['Ingreso'] = df['Ingreso'].where(primeira_ocorrencia, 0)

# Calcular ingresso total
df['ingresso_total'] = df['INGRESODT'] + df['Ingreso']

print(f"Linhas com Ingreso > 0: {(df['Ingreso'] > 0).sum()}")
print(f"\nValidação:")
print(f"Soma INGRESODT: {df['INGRESODT'].sum():,.2f}")
print(f"Soma Ingreso: {df['Ingreso'].sum():,.2f}")
print(f"Soma ingresso_total: {df['ingresso_total'].sum():,.2f}")

df[['REFERENCIA', 'INGRESODT', 'Ingreso', 'ingresso_total']].head(10)
```

```python
# Somar INGRESODT normalmente (é por linha de entrega, isso está correto)
soma_ingresodt = df['INGRESODT'].sum()

# Para ingresso_danone, só contar 1x por REFERENCIA_limpo (evitar duplicar)
ingresso_danone_unico = df.drop_duplicates(subset='REFERENCIA_limpo')['Ingreso'].sum()

print(f"Soma INGRESODT (todas as linhas): {soma_ingresodt:,.2f}")
print(f"Soma Ingreso (1x por REFERENCIA): {ingresso_danone_unico:,.2f}")
print(f"Total correto: {soma_ingresodt + ingresso_danone_unico:,.2f}")

print(f"\n{'='*50}")
print("Comparação com soma errada (linha a linha, com duplicação):")
print(f"Soma Ingreso (todas as linhas, ERRADO): {df['Ingreso'].sum():,.2f}")
print(f"Diferença por duplicação: {df['Ingreso'].sum() - ingresso_danone_unico:,.2f}")
```

```python
# 1. Agregar duplicados de trator+data no km_diario (somar valores numéricos, manter primeiro sub_divisao)
df_km_agg = df_km.groupby(['trator', 'data'], as_index=False).agg({
    'km': 'sum',
    'portagens': 'sum',
    'preco_base': 'sum',
    'sub_divisao': 'first',
    'total': 'sum'
})

print(f"Linhas df_km antes de agregar: {len(df_km)}")
print(f"Linhas df_km após agregar duplicados: {len(df_km_agg)}")

# 2. Preparar chave de join
df_km_agg['data'] = pd.to_datetime(df_km_agg['data']).dt.date
df_final['FCARGA_date'] = pd.to_datetime(df_final['FCARGA']).dt.date

df_final['TRACTORA_limpo'] = df_final['TRACTORA'].str.strip()
df_km_agg['trator_limpo'] = df_km_agg['trator'].str.strip()

# 3. Merge (left, mantendo todas as linhas do df_final)
linhas_antes = len(df_final)

df_final = df_final.merge(
    df_km_agg[['trator_limpo', 'data', 'km', 'portagens', 'preco_base', 'sub_divisao', 'total']],
    left_on=['TRACTORA_limpo', 'FCARGA_date'],
    right_on=['trator_limpo', 'data'],
    how='left'
)

print(f"\nLinhas df_final antes do merge: {linhas_antes}")
print(f"Linhas df_final depois do merge: {len(df_final)}")

# 4. Marcar primeira ocorrência de cada combinação TRACTORA+FCARGA_date (evitar duplicar somas)
primeira_ocorrencia = ~df_final.duplicated(subset=['TRACTORA_limpo', 'FCARGA_date'])

for col in ['km', 'portagens', 'preco_base', 'total']:
    df_final[col] = df_final[col].where(primeira_ocorrencia, 0)

# 5. Limpar colunas auxiliares
df_final = df_final.drop(columns=['TRACTORA_limpo', 'trator_limpo', 'data', 'FCARGA_date'])

print(f"\nLinhas com km preenchido: {(df_final['km'] > 0).sum()} de {len(df_final)}")
print(f"Soma km (df_final): {df_final['km'].sum():,.2f}")
print(f"Soma km (df_km original): {df_km['km'].sum():,.2f}")
```

```python
# Converter tipos de dados numéricos
df['INGRESODT'] = pd.to_numeric(df['INGRESODT'], errors='coerce')
df['COSTEDT'] = pd.to_numeric(df['COSTEDT'], errors='coerce')
df['RENTADT'] = pd.to_numeric(df['RENTADT'], errors='coerce')
df['PALETSDT'] = pd.to_numeric(df['PALETSDT'], errors='coerce')
df['PESO_BRUTO'] = pd.to_numeric(df['PESO_BRUTO'], errors='coerce')
df['PALETS'] = pd.to_numeric(df['PALETS'], errors='coerce')

# CODEUT como string
df['CODEUT'] = df['CODEUT'].astype(str)

# Trim em strings (apenas colunas efetivamente texto, evita erro com colunas date)
def trim_se_string(col):
    if col.dtype == "object" and col.apply(lambda v: isinstance(v, str) or pd.isna(v)).all():
        return col.str.strip()
    return col

df = df.apply(trim_se_string)

# Datas como datetime (sem hora, prontas para PBI) - DEPOIS do trim
df['FCARGA'] = pd.to_datetime(df['FCARGA'], format='%Y%m%d', errors='coerce').dt.date
df['FENTREGA'] = pd.to_datetime(df['FENTREGA'], format='%Y%m%d', errors='coerce').dt.date

# Colunas auxiliares ano, mês e SEMANA
df['ano'] = pd.to_datetime(df['FCARGA']).dt.year
df['mes_num'] = pd.to_datetime(df['FCARGA']).dt.month
df['semana'] = pd.to_datetime(df['FCARGA']).dt.isocalendar().week
df['mes'] = pd.to_datetime(df['FCARGA']).dt.to_period('M')
# Coluna de ordenação para PBI (ordena meses, dias e semanas cronologicamente)
df['ordem_data'] = pd.to_datetime(df['FCARGA']).dt.strftime('%Y%m%d').astype(int)
# Dia da semana
df['dia_semana'] = pd.to_datetime(df['FCARGA']).dt.day_name()


# Ordenação dos dias da semana (domingo = 1)
ordem_dia_semana_map = {
    'Sunday': 1, 'Monday': 2, 'Tuesday': 3, 'Wednesday': 4,
    'Thursday': 5, 'Friday': 6, 'Saturday': 7
}
df['ordem_dia_semana'] = df['dia_semana'].map(ordem_dia_semana_map)


# Corrigir tipos numéricos que ficaram como object (Danone)
df['Ingreso'] = pd.to_numeric(df['Ingreso'], errors='coerce')
df['ingresso_total'] = pd.to_numeric(df['ingresso_total'], errors='coerce')

# Selecionar colunas finais
colunas_finais = [
    'PROPIETARIO', 'TRANSPORTISTA', 'TRACTORA', 'REMOLQUE',
    'INGRESODT', 'COSTEDT', 'RENTADT',
    'PALESREAL' if 'PALESREAL' in df.columns else 'PALETSDT', 'PESO_BRUTO', 'PALETS',
    'CODEUT', 'CODEDT',
    'FCARGA', 'FENTREGA', 'ano', 'mes_num',
    'ACTIVIDAD', 'CODACT', 'REFERENCIA',
    'LOCORIGEN', 'PROV_ORIGEN', 'PAISORIGEN',
    'LOCDESTINO', 'PROV_DESTINO', 'PAISDESTINO',
    'ENTREGAR', 'DESTINO',
    'LOCCAR', 'LUGARCARGA', 'LOCDES', 'LUGARDESCARGA',
    'TIPOCLIENTE', 'TIPOFLUJO', 'TIPOPALETA', 'TEMP_MERC_PED',
    'CAMION_TIPO', 'CAMION_CAPACIDAD',
    'Ingreso', 'ingresso_total',
    'km', 'portagens', 'preco_base', 'sub_divisao', 'total','ordem_data','semana','dia_semana','ordem_dia_semana'
]

df_final = df[[c for c in colunas_finais if c in df.columns]].copy()

print(f"Shape df_final: {df_final.shape}")
print(f"\nTipos de dados:")
print(df_final.dtypes)
print(f"\nNulos por coluna:")
print(df_final.isna().sum()[df_final.isna().sum() > 0])
```
