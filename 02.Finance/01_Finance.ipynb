```python
import sqlite3
import pandas as pd
import platform
import warnings
warnings.filterwarnings('ignore')

# Detecção do SO e caminho BD
if platform.system() == 'Windows':
    DB_PATH = r"C:\Users\LISARR\Documents\python\01.Financeiro\inform_27.db"
elif platform.system() == 'Darwin':
    DB_PATH = "/Volumes/RR/DB/inform_27.db"
else:
    DB_PATH = "inform_27.db"

# Conectar
con = sqlite3.connect(DB_PATH)
cursor = con.cursor()
```

```python
df = pd.read_sql_query("SELECT * FROM inform_26_2026", con)
con.close()
```

```python
print(f"Shape: {df.shape}")
print(f"\nColunas: {df.shape[1]}")
df.head()
```

```python
# Converter tipos de dados
df['INGRESOUT'] = pd.to_numeric(df['INGRESOUT'], errors='coerce')
df['COSTEUT'] = pd.to_numeric(df['COSTEUT'], errors='coerce')
df['MARGENUT'] = pd.to_numeric(df['MARGENUT'], errors='coerce')
df['PALESEQUI'] = pd.to_numeric(df['PALESEQUI'], errors='coerce')
df['OCUPALETS'] = pd.to_numeric(df['OCUPALETS'], errors='coerce')
df['PALESREAL'] = pd.to_numeric(df['PALESREAL'], errors='coerce')
df['PESO_BRUTO'] = pd.to_numeric(df['PESO_BRUTO'], errors='coerce')
df['OCUKILOS'] = pd.to_numeric(df['OCUKILOS'], errors='coerce')
df['KM'] = pd.to_numeric(df['KM'], errors='coerce')
df['EUR_KM'] = pd.to_numeric(df['EUR_KM'], errors='coerce')

# Datas como datetime
df['FCARGA'] = pd.to_datetime(df['FCARGA'], format='%Y%m%d', errors='coerce')
df['FECHA_LLEGADA'] = pd.to_datetime(df['FECHA_LLEGADA'], format='%Y%m%d', errors='coerce')
df['FECHA_PREFACTURA'] = pd.to_datetime(df['FECHA_PREFACTURA'], format='%Y%m%d', errors='coerce')

# CODEUT como string
df['CODEUT'] = df['CODEUT'].astype(str)

# Trim em strings
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
print(f"Shape: {df.shape}")
print(f"\nTipos de dados:")
print(df.dtypes)
print(f"\nNulos por coluna:")
print(df.isnull().sum())
df.head()
```

```python
# Custo por palete
df['custo_palete'] = df['COSTEUT'] / df['PALESREAL']

# Custo por kg
df['custo_kg'] = df['COSTEUT'] / df['PESO_BRUTO']

# Apenas registos com custo > 0 e pales > 0
df_valido = df[(df['COSTEUT'] > 0) & (df['PALESREAL'] > 0) & (df['PESO_BRUTO'] > 0)].copy()

print(f"Registos válidos: {len(df_valido)} de {len(df)}")
```

```python
# Extrair mês de FCARGA
df_valido['mes'] = df_valido['FCARGA'].dt.to_period('M')

# Agrupar por transportador e mês
custo_palete_mes = df_valido.groupby(['TRANSPORTISTA', 'mes']).agg({
    'custo_palete': 'mean',
    'PALESREAL': 'sum'
}).round(2)

# Resetar índice
custo_palete_mes = custo_palete_mes.reset_index()

# Renomear colunas para pivot
custo_palete_mes.columns = ['TRANSPORTISTA', 'mes', 'custo_medio', 'n_paletes']

# Stack para ter (mes, metrica) como colunas
resultado = custo_palete_mes.set_index(['TRANSPORTISTA', 'mes']).stack().unstack([1, 2])
resultado.columns.names = ['mes', 'metrica']

print("Custo Médio por Palete e Nº Paletes por Mês:")
resultado
```
