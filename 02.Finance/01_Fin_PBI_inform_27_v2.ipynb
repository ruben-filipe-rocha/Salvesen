```python
import sqlite3
import pandas as pd
import platform
import warnings
warnings.filterwarnings('ignore')

# Detecção do SO e caminho BD
if platform.system() == 'Windows':
    DB_PATH = r"C:\Users\LISARR\Documents\python\01.Financeiro\inform_27.db"
elif platform.system() == 'Darwin':
    DB_PATH = "/Volumes/RR/DB/inform_27.db"
else:
    DB_PATH = "inform_27.db"

# Conectar
con = sqlite3.connect(DB_PATH)
cursor = con.cursor()
```

```python
df = pd.read_sql_query("SELECT * FROM inform_27_2026", con)
con.close()
```

```python
print(f"Shape: {df.shape}")
print(f"\nColunas: {df.shape[1]}")
df.head()
```

```python
# Converter tipos de dados numéricos
df['INGRESODT'] = pd.to_numeric(df['INGRESODT'], errors='coerce')
df['COSTEDT'] = pd.to_numeric(df['COSTEDT'], errors='coerce')
df['RENTADT'] = pd.to_numeric(df['RENTADT'], errors='coerce')
df['PALETSDT'] = pd.to_numeric(df['PALETSDT'], errors='coerce')
df['PESO_BRUTO'] = pd.to_numeric(df['PESO_BRUTO'], errors='coerce')
df['PALETS'] = pd.to_numeric(df['PALETS'], errors='coerce')
df['KM'] = pd.to_numeric(df['KM'], errors='coerce')
df['KMREALES'] = pd.to_numeric(df['KMREALES'], errors='coerce')

# CODEUT como string
df['CODEUT'] = df['CODEUT'].astype(str)

# Trim em strings (ANTES de converter as datas)
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)

# Datas como datetime (sem hora, prontas para PBI) - DEPOIS do trim
df['FCARGA'] = pd.to_datetime(df['FCARGA'], format='%Y%m%d', errors='coerce').dt.date
df['FENTREGA'] = pd.to_datetime(df['FENTREGA'], format='%Y%m%d', errors='coerce').dt.date
```

```python
print(f"Shape: {df.shape}")
print(f"\nTipos de dados:")
print(df.dtypes)
print(f"\nNulos por coluna:")
print(df.isnull().sum())
df.head()
```

```python
# Caminho do ficheiro
caminho = r"C:\Users\LISARR\Documents\python\01.Financeiro\Danone_Custos_V2.xlsx"

# Ler sheet "Ingreso y coste" colunas 0-4 (A-E) a partir da linha 2
df_danone = pd.read_excel(
    caminho,
    sheet_name="Ingreso y coste",
    usecols=range(5),
    skiprows=1,
    header=None
)

# Renomear colunas
df_danone.columns = ['A', 'B', 'C', 'D', 'E']

print(f"Shape: {df_danone.shape}")
df_danone.head()
```

```python
# Preparar df_danone (renomear colunas)
df_danone.columns = ['CODEDT', 'REFERENCIA', 'CODEUT', 'COSTEDT_DANONE', 'Ingreso']

# Remover primeira linha se for header
if df_danone.iloc[0, 0] == 'CODEDT':
    df_danone = df_danone.iloc[1:].reset_index(drop=True)

# Converter tipos de dados
df_danone['CODEDT'] = pd.to_numeric(df_danone['CODEDT'], errors='coerce').fillna(0).astype('Int64')
df_danone['CODEUT'] = df_danone['CODEUT'].astype(str).str.strip()
df_danone['REFERENCIA'] = df_danone['REFERENCIA'].astype(str).str.strip()
df_danone['Ingreso'] = pd.to_numeric(df_danone['Ingreso'], errors='coerce')

# Limpeza de espaços no df
df['REFERENCIA'] = df['REFERENCIA'].astype(str).str.strip()
df['CODEUT'] = df['CODEUT'].astype(str).str.strip()
df['CODEDT'] = pd.to_numeric(df['CODEDT'], errors='coerce').fillna(0).astype('Int64')

# Função de lookup com cascata - SEM CRITÉRIOS DE FILTRO
def buscar_ingreso_danone(row):
    codeut = str(row['CODEUT']).strip()
    referencia = str(row['REFERENCIA']).strip()
    codedt = int(row['CODEDT']) if pd.notna(row['CODEDT']) else None
    
    # Tentativa 1: CODEUT + REFERENCIA
    match1 = df_danone[(df_danone['CODEUT'] == codeut) & (df_danone['REFERENCIA'] == referencia)]
    if not match1.empty:
        return match1['Ingreso'].iloc[0]
    
    # Tentativa 2: CODEUT + CODEDT
    if codedt is not None:
        match2 = df_danone[(df_danone['CODEUT'] == codeut) & (df_danone['CODEDT'] == codedt)]
        if not match2.empty:
            return match2['Ingreso'].iloc[0]
    
    return None

# Aplicar função
df['ingresso_danone'] = df.apply(buscar_ingreso_danone, axis=1)

# Estatísticas
total_linhas = len(df)
matches = df['ingresso_danone'].notna().sum()
sum_ingresso_danone = df['ingresso_danone'].sum()
sum_ingreso_excel = df_danone['Ingreso'].sum()

print(f"Total de linhas df: {total_linhas}")
print(f"Matches encontrados: {matches}")
print(f"Taxa sucesso: {(matches/total_linhas*100):.2f}%")
print(f"\nSum ingresso_danone: {sum_ingresso_danone:,.2f}")
print(f"Sum Ingreso Excel: {sum_ingreso_excel:,.2f}")
print(f"Diferença: {(sum_ingreso_excel - sum_ingresso_danone):,.2f}")

df[['CODEDT', 'REFERENCIA', 'CODEUT', 'INGRESODT', 'ingresso_danone']].head(10)
```

```python
# Validação: Linhas do Excel que têm/não têm match no df

print("="*70)
print("VALIDAÇÃO - LINHAS DO EXCEL (df_danone)")
print("="*70)

# Total de linhas no Excel
total_danone = len(df_danone)
print(f"\nTotal de linhas no Excel: {total_danone}")

# Função para verificar se linha do Excel tem match no df
def tem_match_no_df(row_danone):
    codeut = str(row_danone['CODEUT']).strip()
    referencia = str(row_danone['REFERENCIA']).strip()
    codedt = int(row_danone['CODEDT']) if pd.notna(row_danone['CODEDT']) else None
    
    # Tentativa 1: CODEUT + REFERENCIA
    match1 = df[(df['CODEUT'] == codeut) & (df['REFERENCIA'] == referencia)]
    if not match1.empty:
        return True
    
    # Tentativa 2: CODEUT + CODEDT
    if codedt is not None:
        match2 = df[(df['CODEUT'] == codeut) & (df['CODEDT'] == codedt)]
        if not match2.empty:
            return True
    
    return False

# Aplicar função
df_danone['tem_match'] = df_danone.apply(tem_match_no_df, axis=1)

# Contar
com_match = df_danone['tem_match'].sum()
sem_match = total_danone - com_match

print(f"\nLinhas com match no df: {com_match}")
print(f"Linhas SEM match no df: {sem_match}")
print(f"Taxa cobertura: {(com_match/total_danone*100):.2f}%")

# Somas dos Ingreso
sum_com_match = df_danone[df_danone['tem_match']]['Ingreso'].sum()
sum_sem_match = df_danone[~df_danone['tem_match']]['Ingreso'].sum()

print(f"\nSum Ingreso (com match): {sum_com_match:,.2f}")
print(f"Sum Ingreso (SEM match): {sum_sem_match:,.2f}")
print(f"Total: {(sum_com_match + sum_sem_match):,.2f}")

print("\n" + "="*70)
print("LINHAS DO EXCEL SEM MATCH NO DF (primeiras 20)")
print("="*70)
df_danone[~df_danone['tem_match']][['CODEDT', 'REFERENCIA', 'CODEUT', 'Ingreso']].head(20)
```

```python
# Número de DT (linhas) por mês

print("="*70)
print("DISTRIBUIÇÃO DE DT POR MÊS")
print("="*70)

# Extrair mês de FCARGA
df['mes'] = pd.to_datetime(df['FENTREGA']).dt.month
df['mes_nome'] = pd.to_datetime(df['FENTREGA']).dt.strftime('%B')

# Contar por mês
dt_por_mes = df['mes'].value_counts().sort_index()

print("\nNúmero de DT por mês:")
for mes, count in dt_por_mes.items():
    mes_nome = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez'][mes-1]
    pct = (count / len(df) * 100)
    print(f"  {mes_nome}: {count:>6} DT ({pct:>5.2f}%)")

print(f"\nTotal: {len(df)} DT")
```
